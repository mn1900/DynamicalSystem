```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Create grid
x = np.linspace(-2.5, 2.5, 500)
y = np.linspace(-2.5, 2.5, 500)
X, Y = np.meshgrid(x, y)

# Derivatives
dx = -Y + X*(1 - X**2 - Y**2)
dy = X + Y*(1 - X**2 - Y**2)

# Nullclines (dashed and thin)
plt.contour(X, Y, dx, levels=[0], colors='blue',
            linestyles='--', linewidths=1)

plt.contour(X, Y, dy, levels=[0], colors='red',
            linestyles='--', linewidths=1)


# System of differential equations
def system(t, z):
    x, y = z
    dx = -y + x*(1 - x**2 - y**2)
    dy = x + y*(1 - x**2 - y**2)
    return [dx, dy]


# Initial conditions
initial_conditions = [
    (0.1, 0.3),
    (0.12, 0.2),
    (-1.7, 0),
    (1.2,-2),
    (0, -0.7)
]

# Plot trajectories (solid and thin)
for initial in initial_conditions:
    solution = solve_ivp(
        system,
        [0, 20],
        initial,
        max_step=0.02
    )

    plt.plot(
        solution.y[0],
        solution.y[1],
        'k-',
        linewidth=1
    )


# Axes
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

plt.xlabel('x')
plt.ylabel('y')

plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)

plt.gca().set_aspect('equal')

plt.show()
```
